```python
raw_score_margin = df.pts - df.opp_pts
score_difference_only = raw_score_margin.abs()
is_game_close = score_difference_only <= 3
df['is_close_game'] = is_game_close 
data_grouped_by_decade = df.groupby('decade')
close_game_proportion = data_grouped_by_decade['is_close_game'].mean()
final_close_game_rate = close_game_proportion * 100

print("Percentage of Games Decided by 3 Points or Less, by Decade:")
print(final_close_game_rate)
```

```python
import matplotlib.pyplot as plt 
final_close_game_rate.plot(
    kind='line',           
    title='Close Games Rate Trend (Decade)' 
)
plt.ylabel('Percentage of Close Games (%)')
plt.show()
```

```python
only_home_games = df[df.game_location == 'H']
only_close_games = df[df.is_close_game == True]

critical_games = df[
    (df.game_location == 'H') & 
    (df.is_close_game == True)
]
critical_games_grouped = critical_games.groupby('decade')
win_proportion = critical_games_grouped['win'].mean()
close_game_home_win_rate = win_proportion * 100

print("Home Win Percentage in Close Games (Margin <= 3 Pts) by Decade:")
print(close_game_home_win_rate)
```

```python
filtered_rate = close_game_home_win_rate.loc[1950:]
filtered_rate.plot(
    kind='line',                   
    title='Home Win Percentage in Close Games (Since 1950)' 
)
plt.ylabel('Home Win Percentage (%)')
plt.show()
```

```python
df['total_score'] = df.pts + df.opp_pts
average_pace = df.groupby('decade')['total_score'].mean()
print(average_pace)
```

```python
df['margin'] = (df.pts - df.opp_pts).abs()

# --- 2. Group and Find the Average Margin of Victory per Decade ---
# We track how much smaller the average margin gets over time.
average_margin = df.groupby('decade')['margin'].mean()

print("Average Margin of Victory by Decade:")
print(average_margin)
```

```python
# --- 1. Calculate the raw ELO difference (Home Team ELO minus Opponent ELO) ---
# ELO is a good proxy for team skill/quality.
df['home_elo_advantage'] = df.elo_n - df.opp_elo_n

# --- 2. Group and Find the Average Home ELO Advantage per Decade ---
# We are seeing if the home team is starting with less of a skill advantage over time.
average_home_elo_advantage = df.groupby('decade')['home_elo_advantage'].mean()

print("Average Pre-Game ELO Advantage for Home Team by Decade:")
print(average_home_elo_advantage)
```

```python
max_elo_by_decade = df.groupby('decade')['elo_n'].max()

print("Maximum ELO Rating Achieved by Any Team, by Decade:")
print(max_elo_by_decade)
```
